```python
#edge_costs.json文件生成（使用output中的第一个构建目标）
import json

with open(r"\\wsl.localhost\Ubuntu\home\enovoleekira\function_test\project\Catch2\edge_costs\build_time_estimation.json", "r") as f:
    data = json.load(f)

cost_map = {}
for entry in data:
    if "outputs" in entry and entry["outputs"]:
        cost_map[entry["outputs"][0]] = entry["estimated_build_time"]

with open(r"\\wsl.localhost\Ubuntu\home\enovoleekira\function_test\project\Catch2\edge_costs\edge_costs.json", "w") as f:
    json.dump(cost_map, f, indent=2)
```
